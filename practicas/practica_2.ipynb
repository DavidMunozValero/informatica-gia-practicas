### Ejercicio 6: Perfil térmico de la Atmósfera Estándar Internacional (ISA)

En la atmósfera estándar internacional (ISA), la temperatura varía con la altitud geopotencial $h$ (en metros) según gradientes térmicos definidos por capas:

| Capa atmosférica | Rango de altitud $h$ (m) | Modelo de temperatura $T(h)$ (°C) |
|---|---|---|

| **Tropopausa / Baja Estratosfera** | $11\,000 < h \le 20\,000$ | $T = -56.5$ (isoterma) |
| **Estratosfera media** | $20\,000 < h \le 32\,000$ | $T = -56.5 + 0.0010 \cdot (h - 20000)$ |
| **Estratosfera superior** | $32\,000 < h \le 47\,000$ | $T = -44.5 + 0.0028 \cdot (h - 32000)$ |
